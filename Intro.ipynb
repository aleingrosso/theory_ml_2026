## Course

How to deal with high-dimensional probability distributions in machine learning, along two complementary routes: **approximation** (Monte Carlo, variational methods, message passing) and **trivialization in high dimension** (concentration of measure and the statistical physics of learning, online learning, infinite-width networks).

### Lectures
1. Monte Carlo methods and spin models
2. Variational approximation
3. Belief propagation
4. Statistical physics of machine learning
5. Online learning in neural networks
6. Kernel methods and infinite-width networks
7. Attention and transformers

### Course material
Each lecture comes with an interactive jupyter notebook; Lectures 4–6 also come with lecture notes (pdf) for the derivations.

Clone the github repository at https://github.com/aleingrosso/theory_ml_2026.

## Evaluation

**Individual oral exam** (100% of the grade), about 30 minutes, with three kinds of questions:

* theoretical questions on the topics of the lectures;
* derivations carried out during the lectures;
* questions on the implementation of a method you worked on in the weekly practicals (Python or pseudocode).

The weekly practicals are **not graded**, but they are important for preparation for the exam: work through all of them.

## Practicals

* Each lecture ends with a set of exercises (*Assignments*).
* Keep your code and notes: at the exam you may be asked to explain your implementation, the choices you made (parameters, initialization, convergence checks) and your results.
* When making plots, be very careful with titles, axis labels and captions, and clarify all parameters used in each plot.
